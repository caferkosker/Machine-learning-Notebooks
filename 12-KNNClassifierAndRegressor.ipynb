```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier,KNeighborsRegressor
```

```python
df = pd.read_csv("12-health_risk_classification.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
sns.scatterplot(x=df["blood_pressure_variation"],y=df["activity_level_index"],hue=df["high_risk_flag"])
plt.show()
```

```python
df["high_risk_flag"].value_counts()
```

```python
X = df.drop("high_risk_flag",axis=1)
y = df["high_risk_flag"]
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.25,random_state=15)
```

```python
sns.boxenplot(df)
plt.show()
```

```python
scaler = StandardScaler()
```

```python
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
classifier =  KNeighborsClassifier(n_neighbors=5,algorithm="auto",weights="uniform")
```

```python
classifier.fit(X_train_scaled,y_train)
```

```python
y_pred = classifier.predict(X_test_scaled)
```

```python
print("Confusion Matrix: \n",confusion_matrix(y_test,y_pred))
print("Classification Report: ",classification_report(y_test,y_pred))
print("Accuracy Score: ",accuracy_score(y_test,y_pred))
```

```python
classifier =  KNeighborsClassifier(n_neighbors=5,algorithm="kd_tree",weights="uniform")
classifier.fit(X_train_scaled,y_train)
```

```python
y_pred = classifier.predict(X_test_scaled)
print("Confusion Matrix: \n",confusion_matrix(y_test,y_pred))
print("Classification Report: ",classification_report(y_test,y_pred))
print("Accuracy Score: ",accuracy_score(y_test,y_pred))
```

```python
classifier =  KNeighborsClassifier(n_neighbors=5,algorithm="ball_tree",weights="uniform")
classifier.fit(X_train_scaled,y_train)
```

```python
y_pred = classifier.predict(X_test_scaled)
print("Confusion Matrix: \n",confusion_matrix(y_test,y_pred))
print("Classification Report: ",classification_report(y_test,y_pred))
print("Accuracy Score: ",accuracy_score(y_test,y_pred))
```

```python
classifier =  KNeighborsClassifier(n_neighbors=3,algorithm="kd_tree",weights="uniform")
classifier.fit(X_train_scaled,y_train)
```

```python
y_pred = classifier.predict(X_test_scaled)
print("Confusion Matrix: \n",confusion_matrix(y_test,y_pred))
print("Classification Report: ",classification_report(y_test,y_pred))
print("Accuracy Score: ",accuracy_score(y_test,y_pred))
```

```python
sns.scatterplot(x=df_reg["outdoor_humidity_level"],y=df_reg["daily_energy_consumption_kwh"])
plt.show()
```

```python
df_reg.corr()
```

```python
X = df_reg.drop("daily_energy_consumption_kwh",axis=1)
y = df_reg["daily_energy_consumption_kwh"]
```

```python
scaler = StandardScaler()
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.25,random_state=15)
```

```python
X_train_scaled_reg = scaler.fit_transform(X_train)
X_test_scaled_reg = scaler.transform(X_test)
```

```python
regressor = KNeighborsRegressor(n_neighbors=5,algorithm="auto")
```

```python
regressor.fit(X_train_scaled_reg,y_train)
```

```python
y_pred_reg = regressor.predict(X_test_scaled_reg)
```

```python
print("Mean Absolute Error: ",mean_absolute_error(y_test,y_pred_reg))
print("Mean Squared Error: ",mean_squared_error(y_test,y_pred_reg))
print("R2 Score: ",r2_score(y_test,y_pred_reg))
```

```python
plt.scatter(y_test,y_pred_reg)
plt.show()
```

```python
regressor = KNeighborsRegressor(n_neighbors=7,algorithm="auto")
regressor.fit(X_train_scaled_reg,y_train)
y_pred_reg = regressor.predict(X_test_scaled_reg)
print("Mean Absolute Error: ",mean_absolute_error(y_test,y_pred_reg))
print("Mean Squared Error: ",mean_squared_error(y_test,y_pred_reg))
print("R2 Score: ",r2_score(y_test,y_pred_reg))
plt.scatter(y_test,y_pred_reg)
plt.show()
```
