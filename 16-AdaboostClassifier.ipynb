```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import AdaBoostClassifier
from sklearn.metrics import classification_report,confusion_matrix,accuracy_score
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import GaussianNB
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
```

```python
columns_to_check = ["Glucose","BloodPressure","SkinThickness","Insulin","BMI"]

for col in columns_to_check:
    zero_count = (df[col] == 0).sum()
    zero_percentage = 100 * zero_count / len(df)
    print(f"{col}: {zero_count}  %{zero_percentage:.2f}")
```

```python
X = df.drop("Outcome",axis=1)
y = df["Outcome"]
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=15)
```

```python
columns_to_fill = ["Glucose","BloodPressure","SkinThickness","Insulin","BMI"]
```

```python
medians = {}
for col in columns_to_fill:
    median_value = X_train[X_train[col] != 0][col].median()
    medians[col] = median_value
    X_train[col] = X_train[col].replace(0, median_value)

for col in columns_to_fill:
    X_test[col] = X_test[col].replace(0, medians[col])
```

```python
X_train.describe()
```

```python
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
ada = AdaBoostClassifier()
ada.fit(X_train,y_train)
y_pred = ada.predict(X_test)
```

```python
print(classification_report(y_test, y_pred))
print(accuracy_score(y_test, y_pred))
print(confusion_matrix(y_test, y_pred))
```

```python
ada = AdaBoostClassifier(n_estimators=200,learning_rate=1)
ada.fit(X_train,y_train)
y_pred = ada.predict(X_test)
print(classification_report(y_test, y_pred))
print(accuracy_score(y_test, y_pred))
print(confusion_matrix(y_test, y_pred))
```

```python
def calculate_model_metrics(true,predicted):
    class_report = classification_report(true,predicted)
    conf_matrix = confusion_matrix(true,predicted)
    score = accuracy_score(true,predicted)
    return class_report,conf_matrix,score
```

```python
models = {
    "Logistic Regression" : LogisticRegression(),
    "Naive Bayes" : GaussianNB(),
    "KNN Classifier" : KNeighborsClassifier(),
    "SVC" : SVC(),
    "Decision Tree Classifier" : DecisionTreeClassifier(),
    "Random Forest Classifier" : RandomForestClassifier()
}
```

```python
for i in range(len(list(models))):
    model = list(models.values())[i]
    model.fit(X_train,y_train)
    y_train_pred = model.predict(X_train)
    y_test_pred = model.predict(X_test)

    model_train_class_rep , model_train_confusion,  model_train_score = calculate_model_metrics(y_train,y_train_pred)
    model_test_class_rep , model_test_confusion,  model_test_score = calculate_model_metrics(y_test,y_test_pred)

    print(list(models.values())[i])
    print("Evalution for Training Set")
    print("Classification Report :", model_train_class_rep)
    print("Confusion Matrix :",model_train_confusion)
    print("Accuracy Score :",model_train_score)

    print("------------------------")
    print(list(models.values())[i])
    print("Evalution for Test Set")
    print("Classification Report :", model_test_class_rep)
    print("Confusion Matrix :",model_test_confusion)
    print("Accuracy Score :",model_test_score)

    print("------------------------")
    print("\n")
    
    
    
```
