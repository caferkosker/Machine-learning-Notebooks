```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("6-bank_customers.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
X = df.drop("subscribed",axis=1)
y = df["subscribed"]
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,random_state=15,train_size=0.3)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
logistic = LogisticRegression()
logistic.fit(X_train,y_train)
y_pred = logistic.predict(X_test)
```

```python
y_pred
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report
```

```python
score = accuracy_score(y_pred,y_test)
print("score: ",score)
print(classification_report(y_pred,y_test))
print("Confusion Matrix: \n", confusion_matrix(y_pred,y_test))
```

```python
#hyperparamater tuning
```

```python
model = LogisticRegression()
```

```python
penalty = ["l1","l2","elasticnet"]
c_values = [100,50,10,5,1,0.5,0,1,0.05,0.01]
solver = ["newton-cg","lbfgs","liblinear","sag","saga","newton-cholesky"]
```

```python
params = dict(penalty = penalty, C = c_values, solver = solver)
```

```python
params
```

```python
#grid search cv
```

```python
from sklearn.model_selection import GridSearchCV,StratifiedKFold
```

```python
cv = StratifiedKFold()
```

```python
grid = GridSearchCV(estimator=model,param_grid=params, cv=cv, scoring="accuracy",n_jobs=-1)
```

```python
grid
```

```python
grid.fit(X_train,y_train)
```

```python
grid.best_params_
```

```python
 grid.best_score_
```

```python
y_pred = grid.predict(X_test)
```

```python
score = accuracy_score(y_pred,y_test)
print("score: ",score)
print(classification_report(y_pred,y_test))
print("Confusion Matrix: \n", confusion_matrix(y_pred,y_test))
```

```python
randomcv = RandomizedSearchCV(estimator=model, param_distributions=params,cv=5, n_iter=10, scoring="accuracy")
```
